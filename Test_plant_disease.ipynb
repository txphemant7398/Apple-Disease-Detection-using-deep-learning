### Part 1: Single-Image Test Code (For test_plant_disease.ipynb)

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# 1. Load your saved MobileNetV2 model
model = tf.keras.models.load_model(r"D:\All Study Stuff\python\tf-env\python\HPCA Crop\apple_disease_model_v2.keras")

# Exact alphabetical order of your 5 classes (indices 0 to 4)
CLASS_NAMES = [
    "Apple___Apple_scab",
    "Apple___Black_rot",
    "Apple___Cedar_apple_rust",
    "Apple___healthy",
    "Blotch_Apple"
]

# 2. Load & preprocess a single test image
image_path = r"D:\All Study Stuff\python\tf-env\python\HPCA Crop\Apple rust.jpg"  # Update with an actual image filename

img = tf.keras.utils.load_img(image_path, target_size=(128, 128))
img_array = tf.keras.utils.img_to_array(img)
img_batch = np.expand_dims(img_array, axis=0)  # Shape becomes (1, 128, 128, 3)

# Note: Do NOT divide by 255 here—mobilenet_v2.preprocess_input is already built inside the model!

# 3. Predict & display result
predictions = model.predict(img_batch)
predicted_index = int(np.argmax(predictions[0]))
predicted_label = CLASS_NAMES[predicted_index]
confidence = float(np.max(predictions[0]) * 100)

plt.imshow(img)
plt.title(f"Prediction: {predicted_label} ({confidence:.2f}%)")
plt.axis("off")
plt.show()
```
